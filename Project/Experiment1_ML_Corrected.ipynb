# Experiment 1: Data Understanding – Loan Approval Prediction

**Dataset:** `ML DATASET.csv`  
**Records:** 20,000 loan applications  
**Target variable:** `Loan_Status` (Approved / Rejected)

### Objective
Understand the structure, attributes, data types, missing values, duplicate records, and distribution of the loan approval target before applying machine learning models.

```python
# If using Google Colab, run this cell and upload the actual file: ML DATASET.csv
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

# Load the actual dataset filename
df = pd.read_csv('ML DATASET.csv')

print('Dataset loaded successfully.')
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])

# 1. First five rows
print('\n1. FIRST FIVE ROWS')
display(df.head())

# 2. Dataset shape
print('\n2. DATASET SHAPE')
print(df.shape)

# 3. Column names
print('\n3. COLUMN NAMES')
print(df.columns.tolist())

# 4. Data types
print('\n4. DATA TYPES')
display(df.dtypes.to_frame('Data Type'))

# 5. Dataset information
print('\n5. DATASET INFORMATION')
df.info()

# 6. Statistical summary of numerical columns
print('\n6. STATISTICAL SUMMARY')
display(df.describe().T)

# 7. Missing values
print('\n7. MISSING VALUES')
missing = df.isnull().sum()
missing_table = pd.DataFrame({
    'Missing_Count': missing,
    'Missing_Percentage': (missing / len(df) * 100).round(2)
})
display(missing_table[missing_table['Missing_Count'] > 0])

# 8. Duplicate rows
print('\n8. DUPLICATE ROWS')
print(df.duplicated().sum())

# 9. Target distribution
print('\n9. LOAN STATUS DISTRIBUTION')
print(df['Loan_Status'].value_counts())

# 10. Target percentage
print('\n10. LOAN STATUS PERCENTAGE')
print((df['Loan_Status'].value_counts(normalize=True) * 100).round(2))
```

```python
import matplotlib.pyplot as plt

# Loan Status Distribution
status_counts = df['Loan_Status'].value_counts()

plt.figure(figsize=(7, 5))
status_counts.plot(kind='bar')
plt.title('Loan Status Distribution')
plt.xlabel('Loan Status')
plt.ylabel('Number of Applications')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Dataset-specific observations
- The dataset contains **20,000 loan applications and 39 columns**.
- `Loan_Status` is the prediction target with two classes: **Approved** and **Rejected**.
- The dataset contains missing values in fields such as `Property_Value`, `Collateral_Value`, `Loan_to_Value`, `Other_Income`, `Investments`, `Bank_Balance`, `Credit_Score`, `Tax_Return_Filed`, and `Interest_Rate`.
- No duplicate rows are present in the supplied dataset.
- Missing values should be handled during the data-preprocessing stage before model training.
